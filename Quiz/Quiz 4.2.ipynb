### Visualisasi Data Bunga Iris Menggunakan Heatmap dan Pairplot 🌸

#### Dataset ini berisi pengukuran morfologi bunga iris yang terdiri dari panjang dan lebar sepal (kelopak bunga) serta petal(mahkota bunga). Data ini digunakan untuk menganalisis hubungan antar variabel numerik menggunakan visualisasi seperti heatmap dan pairplot.

#### 1. Import Data

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

sns.set_theme(style="white", palette="pastel")
plt.rcParams['figure.figsize'] = (8,6)
```

```python
# memasukkan data tanpa download datanya.

df = sns.load_dataset('iris')
df.head()
```

#### 2. Preview Data

```python
df
```

```python
# menambahkan 1 variabel baru
df['total'] = df['sepal_length'] + df['sepal_width'] + df['petal_length'] + df['petal_width']
```

#### Simpan Data dalam Bentuk CSV

```python
df.to_csv('iris_data.csv', index=False)
```

#### 3. Seleksi/Filter Hanya yang Numerik 

```python
# ambil numerik
df_numeric = df.select_dtypes(include='number')

# cek
df_numeric.info()

# visualisasi
sns.heatmap(df_numeric.corr(), annot=True, cmap='RdPu')
sns.pairplot(df_numeric)
```

### 4. Visualisasi Heatmap dan Pairplot

#### -Heatmap

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="RdPu",   # warna pink (kayak bunga 🌸)
    linewidths=0.5,
    linecolor='white'
)

plt.title("Heatmap Korelasi Variabel Bunga Iris", fontsize=14)
plt.show()
```

#### Penjelasan :
Berdasarkan heatmap, terlihat bahwa variabel petal_length dan petal_width memiliki korelasi positif yang sangat kuat. Hal ini menunjukkan bahwa semakin besar ukuran petal, maka dimensi lainnya juga cenderung meningkat. Sementara itu, variabel sepal_width menunjukkan korelasi yang lebih lemah terhadap variabel lain, sehingga kontribusinya dalam menjelaskan hubungan antar variabel relatif kecil.

#### -Pairplot

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")

sns.pairplot(
    df,
    hue="species",   # beda warna tiap jenis bunga
    palette="pastel",
    diag_kind="kde"
)

plt.show()
```

#### Penjelasan :
Visualisasi pairplot menunjukkan adanya pola hubungan antar variabel numerik. Variabel petal_length dan petal_width memberikan pemisahan yang paling jelas antar data, sehingga dapat digunakan sebagai indikator utama dalam membedakan karakteristik bunga. Sebaliknya, variabel sepal_width dan sepal_length menunjukkan distribusi yang lebih menyebar dan saling tumpang tindih.

#### Kesimpulan :
Berdasarkan hasil visualisasi heatmap dan pairplot, dapat disimpulkan bahwa variabel petal memiliki hubungan yang paling kuat dan berperan penting dalam membedakan karakteristik data. Sementara itu, variabel sepal memiliki hubungan yang lebih lemah dan kurang signifikan dalam analisis. Oleh karena itu, variabel petal dapat menjadi fokus utama dalam analisis lebih lanjut.
